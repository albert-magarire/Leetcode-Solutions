```python
# Install dependencies
!pip install kagglehub plotly pandas

# Import libraries
import kagglehub
from kagglehub import KaggleDatasetAdapter
import pandas as pd
import plotly.express as px

# Set the path to the file you’d like to load
# You can specify a file name within the dataset (for example, 'listings.csv')
file_path = "listings.csv"

# Load the latest version of the dataset from Kaggle
df = kagglehub.load_dataset(
    KaggleDatasetAdapter.PANDAS,
    "airbnb/seattle",
    file_path
)

print("First 5 records:")
print(df.head())
```

```python
# Simplify and aggregate relevant columns
df = df[['price', 'latitude', 'longitude', 'neighbourhood_cleansed']]

# Clean up the price column (remove $ and commas)
df['price'] = df['price'].replace('[\$,]', '', regex=True).astype(float)

# Remove extreme outliers for clearer visualization
df = df[df['price'] < 500]
print("Cleaned dataset shape:", df.shape)
```

```python
# Create the interactive geospatial map
fig = px.scatter_mapbox(
    df,
    lat='latitude',
    lon='longitude',
    color='price',
    hover_name='neighbourhood_cleansed',
    hover_data={'latitude': False, 'longitude': False, 'price': True},
    color_continuous_scale='Viridis',
    title='Seattle Airbnb Prices by Location',
    zoom=10,
    height=600
)

# Apply map style and formatting
fig.update_layout(
    mapbox_style="open-street-map",
    margin={"r":0, "t":50, "l":0, "b":0},
    title_font=dict(size=22),
    coloraxis_colorbar=dict(title="Price (USD)")
)

fig.show()
```
